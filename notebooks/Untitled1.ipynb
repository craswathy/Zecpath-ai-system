```python
import pdfplumber
```

```python
def read_pdf(file_path):
    text = ""
    with pdfplumber.open(file_path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    return text

sample_text = read_pdf("C:/Users/ASWATHY/Zecpath ai system/data/hr_resume_2.pdf")
print(sample_text[:500])
```
